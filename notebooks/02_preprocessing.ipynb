Demonstrates the exact TextCleaner and stylistic numeric-feature pipeline used at both training and inference time (src/preprocessing/text_cleaner.py, src/features/numeric_features.py).

```python
import sys
sys.path.insert(0, '..')

from src.preprocessing.text_cleaner import build_default_cleaner
from src.features.numeric_features import extract_stylistic_features
```

```python
examples = [
    "<b>SHOCKING!!!</b> You won't believe this secret at http://example.com — contact tips@example.com now!",
    "City officials released the 2026 budget report following Tuesday's public session.",
]

cleaner = build_default_cleaner()
for text in examples:
    print('RAW:  ', text)
    print('CLEAN:', cleaner.clean(text))
    print('NUMERIC FEATURES:', extract_stylistic_features(text))
    print()
```

Notice the numeric features (uppercase ratio, exclamation count, punctuation count, ...) are computed from the **raw** text — they're captured before cleaning strips punctuation and case, so that signal isn't thrown away.
