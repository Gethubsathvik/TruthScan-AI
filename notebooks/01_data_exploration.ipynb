# 01 — Data Exploration

Loads the raw dataset (falling back to the bundled synthetic sample if you haven't added a real dataset yet) and looks at class balance, text length, and a few examples.

```python
import sys
sys.path.insert(0, '..')

import pandas as pd
import matplotlib.pyplot as plt

from src.data.loader import load_raw_dataset, clean_dataset
from src.utils.config import config
```

```python
# Falls back to the bundled sample dataset if data/raw/ is empty, so this
# notebook always runs — swap raw_dir=None once you've added a real dataset.
raw_dir = config.raw_data_dir
if not any(raw_dir.glob('*.csv')) and not any(raw_dir.glob('*.tsv')):
    print('No dataset found in data/raw — using data/sample/ instead.')
    raw_dir = config.sample_data_dir

df = load_raw_dataset(raw_dir=raw_dir)
df = clean_dataset(df)
df.head()
```

## Class distribution

```python
counts = df['label'].value_counts().rename(index=config.label_map)
counts.plot(kind='bar', color=['#55A868', '#C44E52'])
plt.ylabel('Count')
plt.title('REAL vs FAKE')
plt.show()
counts
```

## Article length distribution (by class)

```python
df['content_len'] = df['content'].str.len()
for label_value, label_name in config.label_map.items():
    subset = df[df['label'] == label_value]['content_len']
    plt.hist(subset, bins=30, alpha=0.6, label=label_name)
plt.xlabel('Character length')
plt.ylabel('Count')
plt.legend()
plt.title('Article length by class')
plt.show()
```
